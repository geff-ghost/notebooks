# Job Predictions Study Dataset

```python
import os
import pandas as pd
import numpy as np
```

## Loading and Cleaning the data

```python
files = [file for file in os.listdir()]

for file in files:
    if file.endswith('.csv'):
        df = pd.read_csv(file)
```

## Visualizing and Analyzing this data

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.head()
```

# Line Plot

```python
sns.lineplot(
    data=df,
    x='experience_years',
    y='salary',
    linewidth=2.5,
    marker='o',
    color='green'
)
plt.title(
    'Salary against Experience Years',
    fontsize=14,
    fontweight='bold',
    pad=5
)

plt.xlabel('Year of Experience', fontsize=11)
plt.ylabel('Salary ($)', fontsize=11)

plt.show()
```
