## 3. Checking before you act

```python
path = "Tentry/Me.txt"

if os.path.exists(path):
    print("File exists - safe to load")
else:
    print("Not found. cwd is:", os.getcwd())

# prevents the FileNotFoundError crashes
```

## 5. Environment variables

```python
# Setting one just for this Python process (won't persist after script ends)
os.environ["MY_TEST_VAR"] = "hello"

print(os.environ.get("MY_TEST_VAR"))
print(os.environ.get("DOES_NOT_EXIST", "default value if missing"))
```

**Note:** .get() with a default - that's safer than os.environ["KEY"], which throws an error if the key doesn't exit.

## 6.Splitting a path apart

```python
path = "data/titanic.csv"

print(os.path.dirname(path))
print(os.path.basename(path))
print(os.path.splitext(path))
```

- Useful when you're writing a script that processes many files and needs to grab just the filename, or swap the extension (e.g.saving a cleaned version as **titanic_cleaned.csv**)
