```python
# This Python 3 environment comes with many helpful analytics libraries installed
# It is defined by the kaggle/python Docker image: https://github.com/kaggle/docker-python
# For example, here's several helpful packages to load

import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

# Input data files are available in the read-only "../input/" directory
# For example, running this (by clicking run or pressing Shift+Enter) will list all files under the input directory

import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

# You can write up to 20GB to the current directory (/kaggle/working/) that gets preserved as output when you create a version using "Save & Run All" 
# You can also write temporary files to /kaggle/temp/, but they won't be saved outside of the current session

# Use the kagglehub client library to attach Kaggle resources like competitions, datasets, and models to your session
# Learn more about kagglehub: https://github.com/Kaggle/kagglehub/blob/main/README.md

import kagglehub
# kagglehub.dataset_download('<owner>/<dataset-slug>')
```

```python
df_categories = pd.read_csv("/kaggle/input/datasets/hyerdrac/retail-data/retail_data/categories.csv")
df_customers = pd.read_csv("/kaggle/input/datasets/hyerdrac/retail-data/retail_data/customers.csv")
df_order_details = pd.read_csv("/kaggle/input/datasets/hyerdrac/retail-data/retail_data/order_details.csv")
df_orders = pd.read_csv("/kaggle/input/datasets/hyerdrac/retail-data/retail_data/orders.csv")
df_products = pd.read_csv("/kaggle/input/datasets/hyerdrac/retail-data/retail_data/products.csv")

df_master = (
    df_order_details.merge(df_orders, on="OrderID", how="left")
    .merge(df_customers, on="CustomerID", how="left")
    .merge(df_products, on="ProductID", how="left")
    .merge(df_categories, on="CategoryID", how="left")
)

# Some additional calculation you can do with the data
df_master["NetUnitPrice"] = (
    df_master["UnitPrice"] * (1 - df_master["DiscountRate"])
).round(2)
df_master["TotalRevenue"] = (
    df_master["Quantity"] * df_master["NetUnitPrice"]
).round(2)
df_master["TotalCost"] = (df_master["Quantity"] * df_master["UnitCost"]).round(
    2
)
df_master["Profit"] = (
    df_master["TotalRevenue"] - df_master["TotalCost"]
).round(2)

df_master["RealizedRevenue"] = df_master.apply(
    lambda row: 0.00 if row["IsReturned"] == 1 else row["TotalRevenue"], axis=1
)
df_master["RealizedProfit"] = df_master.apply(
    lambda row: 0.00 if row["IsReturned"] == 1 else row["Profit"], axis=1
)

print(df_master.head(10))
```

# Testing Data Quality with Some Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## Barplots

```python
cat_revenue = (
    df_master.groupby("CategoryName")["RealizedRevenue"]
    .sum()
    .reset_index()
    .sort_values(by="RealizedRevenue", ascending=False)
)

plt.figure(figsize=(10, 5))
ax1 = sns.barplot(
    data=cat_revenue,
    x="RealizedRevenue",
    y="CategoryName",
    palette="Blues_r",
    hue="CategoryName",
    legend=False,
)

plt.title(
    "Total Realized Revenue by Category ($)",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Total Revenue ($)", fontsize=11)
plt.ylabel("Category", fontsize=11)

for p in ax1.patches:
    width = p.get_width()
    ax1.annotate(
        f"${width:,.0f}",
        (width, p.get_y() + p.get_height() / 2.0),
        ha="left",
        va="center",
        xytext=(5, 0),
        textcoords="offset points",
        fontsize=9,
    )

plt.tight_layout()
plt.show()
```

```python
segment_avg = (
    df_master.groupby("CustomerSegment")["TotalRevenue"]
    .mean()
    .reset_index()
    .sort_values(by="TotalRevenue", ascending=False)
)

plt.figure(figsize=(8, 4.5))
ax2 = sns.barplot(
    data=segment_avg,
    x="CustomerSegment",
    y="TotalRevenue",
    palette="Greens_r",
    hue="CustomerSegment",
    legend=False,
)

plt.title(
    "Average Order Value by Customer Segment ($)",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Customer Segment", fontsize=11)
plt.ylabel("Average Basket Value ($)", fontsize=11)

for p in ax2.patches:
    height = p.get_height()
    ax2.annotate(
        f"${height:.2f}",
        (p.get_x() + p.get_width() / 2.0, height),
        ha="center",
        va="bottom",
        xytext=(0, 3),
        textcoords="offset points",
        fontsize=10,
        fontweight="bold",
    )

plt.tight_layout()
plt.show()
```

```python
region_segment_profit = (
    df_master.groupby(["Region", "CustomerSegment"])["RealizedProfit"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(11, 5.5))
sns.barplot(
    data=region_segment_profit,
    x="Region",
    y="RealizedProfit",
    hue="CustomerSegment",
    palette="Set2",
)

plt.title(
    "Average Realized Profit by Region and Customer Segment ($)",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Geographic Region", fontsize=11)
plt.ylabel("Average Profit ($)", fontsize=11)
plt.legend(title="Customer Segment", frameon=True)

plt.tight_layout()
plt.show()
```

```python
df_returns = df_master[df_master["IsReturned"] == 1]
return_counts = (
    df_returns.groupby("CategoryName")["IsReturned"]
    .sum()
    .reset_index()
    .sort_values(by="IsReturned", ascending=False)
)

plt.figure(figsize=(10, 5))
ax4 = sns.barplot(
    data=return_counts,
    x="CategoryName",
    y="IsReturned",
    palette="Reds_r",
    hue="CategoryName",
    legend=False,
)

plt.title(
    "Total Returned Units by Product Category",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Category", fontsize=11)
plt.ylabel("Returned Quantity (Units)", fontsize=11)
plt.xticks(rotation=45, ha="right")

for p in ax4.patches:
    height = p.get_height()
    ax4.annotate(
        f"{int(height)}",
        (p.get_x() + p.get_width() / 2.0, height),
        ha="center",
        va="bottom",
        xytext=(0, 3),
        textcoords="offset points",
        fontsize=9,
    )

plt.tight_layout()
plt.show()
```

```python
age_bins = [17, 25, 50, 100]
age_labels = ["Young (18-25)", "Middle-Aged (26-50)", "Senior (51+)"]
df_master["AgeGroup"] = pd.cut(
    df_master["Age"], bins=age_bins, labels=age_labels
)

age_discount = (
    df_master.groupby("AgeGroup", observed=False)["DiscountRate"]
    .mean()
    .reset_index()
    .sort_values(by="DiscountRate", ascending=False)
)

plt.figure(figsize=(7, 4.5))
ax5 = sns.barplot(
    data=age_discount,
    x="AgeGroup",
    y="DiscountRate",
    palette="Purples_r",
    hue="AgeGroup",
    legend=False,
)

plt.title(
    "Average Discount Rate Applied by Age Group",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Age Group", fontsize=11)
plt.ylabel("Average Discount Rate", fontsize=11)

for p in ax5.patches:
    height = p.get_height()
    ax5.annotate(
        f"{height*100:.1f}%",
        (p.get_x() + p.get_width() / 2.0, height),
        ha="center",
        va="bottom",
        xytext=(0, 3),
        textcoords="offset points",
        fontsize=10,
        fontweight="bold",
    )

plt.tight_layout()
plt.show()
```

## Line Plots

```python
df_master["OrderDate"] = pd.to_datetime(df_master["OrderDate"], errors="coerce")
df_master["YearMonth"] = df_master["OrderDate"].dt.to_period("M").dt.to_timestamp()
df_master["OrderYear"] = df_master["OrderDate"].dt.year
```

```python
monthly_financials = (
    df_master.groupby("YearMonth")[["RealizedRevenue", "TotalCost"]]
    .sum()
    .reset_index()
)

plt.figure(figsize=(11, 5))
sns.lineplot(
    data=monthly_financials,
    x="YearMonth",
    y="RealizedRevenue",
    label="Realized Revenue ($)",
    color="darkgreen",
    linewidth=2.5,
)
sns.lineplot(
    data=monthly_financials,
    x="YearMonth",
    y="TotalCost",
    label="Total Cost ($)",
    color="crimson",
    linewidth=2,
    linestyle="--",
)

plt.title(
    "Monthly Revenue vs Cost Trend (Inflation Dynamic)",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Order Timeline", fontsize=11)
plt.ylabel("Amount ($)", fontsize=11)
plt.legend(frameon=True)

plt.tight_layout()
plt.show()
```

```python
category_price_trend = (
    df_master.groupby(["YearMonth", "CategoryName"])["UnitPrice"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=category_price_trend,
    x="YearMonth",
    y="UnitPrice",
    hue="CategoryName",
    palette="tab10",
    linewidth=1.8,
)

plt.title(
    "Average Unit Price Inflation Trajectory by Category",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Order Timeline", fontsize=11)
plt.ylabel("Average Unit Price ($)", fontsize=11)
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left", title="Category")

plt.tight_layout()
plt.show()
```

```python
monthly_discount = (
    df_master.groupby("YearMonth")["DiscountRate"].mean().reset_index()
)

plt.figure(figsize=(11, 4.5))
sns.lineplot(
    data=monthly_discount,
    x="YearMonth",
    y="DiscountRate",
    color="purple",
    linewidth=2,
    marker="o",
    markersize=4,
)

plt.title(
    "Seasonal Discount Rate Fluctuation Timeline",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Order Timeline", fontsize=11)
plt.ylabel("Average Discount Rate", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
segment_yearly_profit = (
    df_master.groupby(["OrderYear", "CustomerSegment"])["RealizedProfit"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(9, 5))
sns.lineplot(
    data=segment_yearly_profit,
    x="OrderYear",
    y="RealizedProfit",
    hue="CustomerSegment",
    marker="s",
    linewidth=2.5,
    palette="Dark2",
)

plt.title(
    "Annual Realized Profit Progression by Customer Segment",
    fontsize=14,
    pad=15,
    fontweight="bold",
)
plt.xlabel("Year", fontsize=11)
plt.ylabel("Total Profit ($)", fontsize=11)
plt.legend(title="Customer Segment")

plt.tight_layout()
plt.show()
```
