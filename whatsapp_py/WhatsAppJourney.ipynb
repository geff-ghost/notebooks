## Indexing, Slicing, Boolean Indexing and Filtering
- One of the biggest advantages of NumPy is how quickly you can access and filter data.

### Array Indexing
- Indexing is used to access a specific element in an array.

```python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50, 60])

# Indexing starts from 0
print(numbers[0])
print(numbers[3])
```

### Array Slicing
- Slicing extracts a portion of an array

```python
print(numbers[1:5])
print(numbers[:3])
print(numbers[3:])
print(numbers[-2:])
```

### Indexing in 2D Arrays
- Access elements using row and column indexes.

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix[1, 2])
```

```python
# Access an entire row:
print(matrix[0])

# Access an entire column
print(matrix[:, 1])
```

### Boolean Indexing
- Boolean indexing selects elements that satisfy a condition.

```python
numbers = np.array([12, 25, 8, 40, 15])
print(numbers > 20)
```

### Filtering Data
- Use Boolean indexing to filter values.

```python
print(numbers[numbers > 20])
```

```python
print(numbers[numbers % 2 == 0])
```

### Multiple Conditions
- Filter using more than one condition.

```python
numbers = np.array([12, 25, 8, 40, 15, 30])
result = numbers[(numbers >= 15) & (numbers <= 30)]
print(result)
```

## Sorting, Searching, Stacking & Spliting Arrays  

### Sorting Arrays
- Numpy provides the **sort()** functions to sort array elements in ascending order.

```python
import numpy as np

numbers = np.array([40, 10, 30, 20])

print(np.sort(numbers))
```

### Searching Elements
- Use **where()** to find the index of an element.

```python
numbers = np.array([10, 20, 30, 20, 40])

print(np.where(numbers == 20))
```

This means the value 20 is present at index 1 and 3.

### Stacking Arrays Vertically
- Combine arrays column-wise using **vstack()**.

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

# combines two arrays row-wise
print(np.vstack((array1, array2)))
```

### Stacking Arrays Horizontally
- Combine arrays column-wise using **hstack()**.

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print(np.hstack((array1, array2)))
```

### Splitting Arrays
- Use **split()** to divide an array into equal parts.

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

print(np.split(numbers, 3))
```

### Copy vs View
#### Copy
- Changes to the original array do not affect the copied array.

```python
numbers = np.array([10, 20, 30])

copy_array = numbers.copy()

numbers[0] = 100

print(copy_array)
print(numbers)
```

#### View
- A view shares the same data as the original array.

```python
numbers = np.array([10, 20, 30])

view_array = numbers.view()

numbers[0] = 100
print(view_array)
```

### Find Unique Values
- Use **unique()** to remove duplicate values.

```python
numbers = np.array([10, 20, 20, 30, 30, 40])
print(np.unique(numbers))
```

```python
# Find Index of a specific value
print(np.where(numbers == 20))
```

# File Handling & Exception Handling
Real-world applications need to store and manage data. That's where file handling becomes important.

## What is File Handling?  

- File handling allows Python to:
1. Read data from files
2. Write data into files
3. Store information permenently

## Opening a File
Use **open()** to work with files.

> Example:
file = open('data.txt', 'r')

- **'r'** means read mode

## Reading a File

```python
file = open('data.txt', 'r')
content = file.read()
print(content)
file.close()
```

## Read Line by Line

```python
file = open('data.txt', 'r')
for line in file:
    print(line)
file.close()
```

## Writing to a File
- Use **'w'** mode to write data.

```python
file = open('notes.txt', 'w')
file.write('Python File Handling')
file.close() # 'w' overwrites existing data
```

## Appending Data
- Use **'a'** mode to add data without deleting old content

```python
file = open('notes.txt', 'a')
file.write('\nNew Line Added')
file.close()
```

## Using **with** Statement
Best way to handle files
1. Automatically closes the file
2. Cleaner and safer code

```python
with open('data.txt', 'r') as file:
    content = file.read()
    print(content)
```

## Working with CSV Files and Exception Handling
- CSV files are used for tabular data.

### What is Exception Handling?
- Errors can crash programs.
- Exception handling prevents that

```python
import csv
try:
    with open('data.csv', 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)
except:
    print('File does not exist!!!')
```

```python
# Handling Specific Errors
try:
    result = 10 / 0
except ZeroDivisionError:
    print('Cannot divide by zero')
```

## Finally Block
- **finally** always runs whether error occurs or not.

```python
try:
    print('Start')
except:
    print('Error')
finally:
    print('Finished')
```

# Object-Orientend Programming (OOP)
- Object-Oriented Programming (OOP) in one of the most important concepts in Python.
- Almost every real world Python application uses OOP concepts.

## What is OOP?
- OOP is a programming approach that organizes code into objects
  
**Each object has:**
- Properties (Data)
- Behavior (Fuctions)

## What is a Class?
- A class is a blueprint for creating objects

**Example:**
class Car:

- Here, 'Car' is a class

## Creating an Object
- An object is an instance of a class:

```python
class Car:
    pass

car1 = Car()
print(type(car1))
```

## Attributes in a Class
- Attributes store information

```python
class Car:
    brand = 'Toyota'

car1 = Car()
print(car1.brand)
```

## Constructor ('init')
- A constructor runs automatically when an object is created.

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

car1 = Car('Toyota')
print(car1.brand)
```

## Instance Methods
- Methods define object behavior

```python
class Car:
    def start(self):
        print('Car Started')

car1 = Car()
car1.start()
```

## Inheritance
- Inheritance allows one class to use features of another class.

```python
class Vehicle:
    def moves(self):
        print('Vehicle Moving')

class Car(Vehicle):
    pass

car1 = Car()
car1.moves()
```

## Encapsulation
- Encapsulation hides data from direct access.

```python
class Account:
    def __init__(self):
        self.__balance = 1000 # private variable

account = Account()
```

## Polymorphism
- Same method name, different behavior.

```python
class Bird:
    def sound(self):
        print('Bird Sound')

class Dog:
    def sound(self):
        print('Dog Sound')

bird = Bird()
dog = Dog()

bird.sound()
dog.sound()
```

## Method Overriding
- Child class can redefine a parent method.

```python
class Vehicle:
    def moves(self):
        print('Vehicle Moving')

class Bike(Vehicle):
    def moves(self):
        print('Bike Moving')

bike = Bike()
bike.moves()
```
