## Indexing, Slicing, Boolean Indexing and Filtering
- One of the biggest advantages of NumPy is how quickly you can access and filter data.

### Array Indexing
- Indexing is used to access a specific element in an array.

```python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50, 60])

# Indexing starts from 0
print(numbers[0])
print(numbers[3])
```

### Array Slicing
- Slicing extracts a portion of an array

```python
print(numbers[1:5])
print(numbers[:3])
print(numbers[3:])
print(numbers[-2:])
```

### Indexing in 2D Arrays
- Access elements using row and column indexes.

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix[1, 2])
```

```python
# Access an entire row:
print(matrix[0])

# Access an entire column
print(matrix[:, 1])
```

### Boolean Indexing
- Boolean indexing selects elements that satisfy a condition.

```python
numbers = np.array([12, 25, 8, 40, 15])
print(numbers > 20)
```

### Filtering Data
- Use Boolean indexing to filter values.

```python
print(numbers[numbers > 20])
```

```python
print(numbers[numbers % 2 == 0])
```

### Multiple Conditions
- Filter using more than one condition.

```python
numbers = np.array([12, 25, 8, 40, 15, 30])
result = numbers[(numbers >= 15) & (numbers <= 30)]
print(result)
```

## Sorting, Searching, Stacking & Spliting Arrays

```python
import numpy as np

numbers = np.array([40, 10, 30, 20])

print(np.sort(numbers))
```

### Stacking Arrays Vertically
- Combine arrays column-wise using **vstack()**.

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

# combines two arrays row-wise
print(np.vstack((array1, array2)))
```

### Stacking Arrays Horizontally
- Combine arrays column-wise using **hstack()**.

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print(np.hstack((array1, array2)))
```

### Splitting Arrays
- Use **split()** to divide an array into equal parts.

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

print(np.split(numbers, 3))
```

### Copy vs View
#### Copy
- Changes to the original array do not affect the copied array.

```python
numbers = np.array([10, 20, 30])

copy_array = numbers.copy()

numbers[0] = 100

print(copy_array)
print(numbers)
```

#### View
- A view shares the same data as the original array.

```python
numbers = np.array([10, 20, 30])

view_array = numbers.view()

numbers[0] = 100
print(view_array)
```

### Find Unique Values
- Use **unique()** to remove duplicate values.

```python
numbers = np.array([10, 20, 20, 30, 30, 40])
print(np.unique(numbers))
```

```python
# Find Index of a specific value
print(np.where(numbers == 20))
```

# File Handling & Exception Handling
Real-world applications need to store and manage data. That's where file handling becomes important.

## What is File Handling?  

- File handling allows Python to:
1. Read data from files
2. Write data into files
3. Store information permenently

## Opening a File
Use **open()** to work with files.

> Example:
file = open('data.txt', 'r')

- **'r'** means read mode

## Reading a File

```python
file = open('data.txt', 'r')
content = file.read()
print(content)
file.close()
```

## Read Line by Line

```python
file = open('data.txt', 'r')
for line in file:
    print(line)
file.close()
```

## Writing to a File
- Use **'w'** mode to write data.

```python
file = open('notes.txt', 'w')
file.write('Python File Handling')
file.close() # 'w' overwrites existing data
```

## Appending Data
- Use **'a'** mode to add data without deleting old content

```python
file = open('notes.txt', 'a')
file.write('\nNew Line Added')
file.close()
```

## Using **with** Statement
Best way to handle files
1. Automatically closes the file
2. Cleaner and safer code

```python
with open('data.txt', 'r') as file:
    content = file.read()
    print(content)
```

## Working with CSV Files and Exception Handling
- CSV files are used for tabular data.

### What is Exception Handling?
- Errors can crash programs.
- Exception handling prevents that

```python
import csv
try:
    with open('data.csv', 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)
except:
    print('File does not exist!!!')
```

```python
# Handling Specific Errors
try:
    result = 10 / 0
except ZeroDivisionError:
    print('Cannot divide by zero')
```

## Finally Block
- **finally** always runs whether error occurs or not.

```python
try:
    print('Start')
except:
    print('Error')
finally:
    print('Finished')
```

# Object-Orientend Programming (OOP)
- Object-Oriented Programming (OOP) in one of the most important concepts in Python.
- Almost every real world Python application uses OOP concepts.

## What is OOP?
- OOP is a programming approach that organizes code into objects
  
**Each object has:**
- Properties (Data)
- Behavior (Fuctions)

## What is a Class?
- A class is a blueprint for creating objects

**Example:**
class Car:

- Here, 'Car' is a class

## Creating an Object
- An object is an instance of a class:

```python
class Car:
    pass

car1 = Car()
print(type(car1))
```

## Attributes in a Class
- Attributes store information

```python
class Car:
    brand = 'Toyota'

car1 = Car()
print(car1.brand)
```

## Constructor ('init')
- A constructor runs automatically when an object is created.

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

car1 = Car('Toyota')
print(car1.brand)
```

## Instance Methods
- Methods define object behavior

```python
class Car:
    def start(self):
        print('Car Started')

car1 = Car()
car1.start()
```

## Inheritance
- Inheritance allows one class to use features of another class.

```python
class Vehicle:
    def moves(self):
        print('Vehicle Moving')

class Car(Vehicle):
    pass

car1 = Car()
car1.moves()
```

## Encapsulation
- Encapsulation hides data from direct access.

```python
class Account:
    def __init__(self):
        self.__balance = 1000 # private variable

account = Account()
```

## Polymorphism
- Same method name, different behavior.

```python
class Bird:
    def sound(self):
        print('Bird Sound')

class Dog:
    def sound(self):
        print('Dog Sound')

bird = Bird()
dog = Dog()

bird.sound()
dog.sound()
```

## Method Overriding
- Child class can redefine a parent method.

```python
class Vehicle:
    def moves(self):
        print('Vehicle Moving')

class Bike(Vehicle):
    def moves(self):
        print('Bike Moving')

bike = Bike()
bike.moves()
```

# Data Cleaning, Missing Values & Duplicates
- Real-world datasets are rarely perfect.
- Before analyzing data, you need to clean in. Data cleaning is one of the most important tasks for every Data Analyst and Data Scientist.

## What are Missing Values?
- Missing values occur when some data is unavailable.

```python
import pandas as pd

data = {
    'Product': ['Laptop','Mobile', None, 'Tablet'],
    'Price': [50000, None, 18000, 15000]
}

df = pd.DataFrame(data)

print(df)
```

- 'NaN' means Not a Number, indicating a missing value

## Detect Missing Values
- Use **isnull()** to identify missing values.

```python
print(df.isnull())
```

```python
# Count missing values in each column
print(df.isnull().sum())
```

## Remove Missing Values
- Use **dropna()** to remove rows with missing values.

```python
clean_df = df.dropna()
print(clean_df)
```

## Fill Missing Values
- Instead of deleting rows, you can replace missing values.

```python
# Replace missing prices with 0
df['Price'] = df['Price'].fillna(0)

# Replace missing product names
df['Product'] = df['Product'].fillna('Unknown')

print(df)
```

## Find Duplicate Rows
- Use **duplicate()** to identify duplicat records.

```python
print(df.duplicated())
```

- If a row is repeated, it wil return **True**

## Remove Duplicate Rows
- Use **drop_duplicates()**

```python
df = df.drop_duplicates()
print(df)
```

- This keeps only unique records

## Check Data Information
- Use **info()** to understand your dataset.

```python
print(df.info())
```

## Generate Summary Statistics
- Use **describe()** to get statistical information.

```python
print(df.describe())
```

**Pro Tip**
- A good data analyst always follows this order:
1. Load the dataset
2. Check **head()**
3. Check **info()**
4. Find missing values
5. Remove duplicates
6. Start the analysis

# GroupBy, Aggregation & Data Analysis
- One of the biggest strengths of Pandas is its ability to summarize large datasets with just a few lines of code.

## Create a Sample DataFrame
- We'll use this dataset throughout the examples

```python
import pandas as pd

data = {
    'Department': ['Sales', 'IT', 'Sales', 'HR', 'IT', 'HR'],
    'Salary': [50000, 70000, 55000, 45000, 75000, 48000],
    'Experience': [2, 5, 3, 1, 6, 2]
}

df = pd.DataFrame(data)
print(df)
```

## What is GroupBy?
- **groupby()** groups rows having the same value.

```python
print(df.groupby('Department'))
```

- This creates groups based on the Department column.

## Calculate Average Salary
- Find the average salary for each department

```python
print(df.groupby('Department')['Salary'].mean())
```

## Calculate Average Salary
- Find the average salary for each department

```python
print(df.groupby('Department')['Salary'].sum())
```

## Multiple Aggregation Functions
- Calculate multiple statistics at once.

```python
print(
    df.groupby('Department')['Salary'].agg(['sum','mean', 'max','min', 'std'])
)
```

## Count Records
- Count the number of records in each department

```python
print(df.groupby('Department').size())
```

## Count Unique Values
- Use **value_counts()** to count occurrences.

```python
print(df['Department'].value_counts())
```

## Find Unique Values
- Display unique department names.

```python
print(df['Department'].unique())
```

## Sort Aggregated Results
- Sort departments by average salary.

```python
result = df.groupby('Department')['Salary'].mean()
print(result.sort_values(ascending=False))
```

## Practice Exercises

```python
# Find the average salary for each department
print(df.groupby('Department')['Salary'].mean())
```

# Merge, Join, Concatenate & Pivot Tables with Examples
- In real-world projects, your data is often spread across multiple files or tables.
- Pandas provide powerful functions to combine and summarize data efficiently.

- In this part, you'll learn:
1. Merge DataFrames
2. Join DataFrames
3. Concatenate DataFrames
4. Pivot Tables
5. CrossTab
6. Real-World Use Cases

## Create Sample DataFrames
- We'll use these DataFrames throughout the examples.

```python
import pandas as pd

products = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'Product': ['Laptop', 'Mobile', 'Tablet']
})

sales = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'Sales': [120, 250, 180]
})
```

## Merge DataFrames
- Use "merge()" to combine DataFrames based on a common column.

```python
result = pd.merge(products, sales, on='ProductID')
print(result)
```

## Join DataFrames
- Use "join()" to combine DataFrames using their indexes.

```python
result = products.join(sales['Sales'])
print(result)
```

## Concatenate DataFrames
- Use "concat()" to combine DataFrames vertically.

```python
first = pd.DataFrame({
    'Value': [10, 20]
})
second = pd.DataFrame({
    'Value': [30, 40]
})
result = pd.concat([first, second])
print(result)
```

## Create a Pivot Table
- A Pivot Table summarizes data quickly.

```python
data = pd.DataFrame({
    'Department': ['Sales', 'Sales', 'IT', 'IT'],
    'Salary': [50000, 55000, 70000, 75000]
})
pivot = data.pivot_table(
    values='Salary',
    index='Department',
    aggfunc='mean'
)
print(pivot)
```

## Create a Crosstab
- A crosstab counts the frequency of combinations between columns.

```python
data = pd.DataFrame({
    'Department': ['Sales', 'Sales', 'IT', 'HR'],
    'Shift': ['Morning', 'Evening', 'Morning', 'Evening']
})
print(pd.crosstab(data['Department'],data['Shift']))
```

**Pro Tip**
- Remember this rule:
1. 'merge()' - Combine using common columns
2. 'join()' - Combine using indexes
3. 'concat()' - Stack DataFrame vertically
4. 'pivot_table()' - Summarize data
5. 'crosstab()' - Count combinations

```python
df1 = pd.DataFrame({'A': [1, 2]})
df2 = pd.DataFrame({'B': [3, 4]})

print(pd.concat([df1, df2]))
```
