# File Handling & Exception Handling
Real-world applications need to store and manage data. That's where file handling becomes important.

## What is File Handling?  

- File handling allows Python to:
1. Read data from files
2. Write data into files
3. Store information permenently

## Opening a File
Use **open()** to work with files.

> Example:
file = open('data.txt', 'r')

- **'r'** means read mode

## Reading a File

```python
file = open('data.txt', 'r')
content = file.read()
print(content)
file.close()
```

## Read Line by Line

```python
file = open('data.txt', 'r')
for line in file:
    print(line)
file.close()
```

## Writing to a File
- Use **'w'** mode to write data.

```python
file = open('notes.txt', 'w')
file.write('Python File Handling')
file.close() # 'w' overwrites existing data
```

## Appending Data
- Use **'a'** mode to add data without deleting old content

```python
file = open('notes.txt', 'a')
file.write('\nNew Line Added')
file.close()
```

## Using **with** Statement
Best way to handle files
1. Automatically closes the file
2. Cleaner and safer code

```python
with open('data.txt', 'r') as file:
    content = file.read()
    print(content)
```

## Working with CSV Files and Exception Handling
- CSV files are used for tabular data.

### What is Exception Handling?
- Errors can crash programs.
- Exception handling prevents that

```python
import csv
try:
    with open('data.csv', 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)
except:
    print('File does not exist!!!')
```

```python
# Handling Specific Errors
try:
    result = 10 / 0
except ZeroDivisionError:
    print('Cannot divide by zero')
```

## Finally Block
- **finally** always runs whether error occurs or not.

```python
try:
    print('Start')
except:
    print('Error')
finally:
    print('Finished')
```

# Object-Orientend Programming (OOP)
- Object-Oriented Programming (OOP) in one of the most important concepts in Python.
- Almost every real world Python application uses OOP concepts.

## What is OOP?
- OOP is a programming approach that organizes code into objects
  
**Each object has:**
- Properties (Data)
- Behavior (Fuctions)

## What is a Class?
- A class is a blueprint for creating objects

**Example:**
class Car:

- Here, 'Car' is a class

## Creating an Object
- An object is an instance of a class:

```python
class Car:
    pass

car1 = Car()
print(type(car1))
```

## Attributes in a Class
- Attributes store information

```python
class Car:
    brand = 'Toyota'

car1 = Car()
print(car1.brand)
```

## Constructor ('init')
- A constructor runs automatically when an object is created.

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

car1 = Car('Toyota')
print(car1.brand)
```

## Instance Methods
- Methods define object behavior

```python
class Car:
    def start(self):
        print('Car Started')

car1 = Car()
car1.start()
```

## Inheritance
- Inheritance allows one class to use features of another class.

```python
class Vehicle:
    def moves(self):
        print('Vehicle Moving')

class Car(Vehicle):
    pass

car1 = Car()
car1.moves()
```

## Encapsulation
- Encapsulation hides data from direct access.

```python
class Account:
    def __init__(self):
        self.__balance = 1000 # private variable

account = Account()
```

## Polymorphism
- Same method name, different behavior.

```python
class Bird:
    def sound(self):
        print('Bird Sound')

class Dog:
    def sound(self):
        print('Dog Sound')

bird = Bird()
dog = Dog()

bird.sound()
dog.sound()
```

## Method Overriding
- Child class can redefine a parent method.

```python
class Vehicle:
    def moves(self):
        print('Vehicle Moving')

class Bike(Vehicle):
    def moves(self):
        print('Bike Moving')

bike = Bike()
bike.moves()
```

# Data Cleaning, Missing Values & Duplicates
- Real-world datasets are rarely perfect.
- Before analyzing data, you need to clean in. Data cleaning is one of the most important tasks for every Data Analyst and Data Scientist.

## What are Missing Values?
- Missing values occur when some data is unavailable.

```python
import pandas as pd

data = {
    'Product': ['Laptop','Mobile', None, 'Tablet'],
    'Price': [50000, None, 18000, 15000]
}

df = pd.DataFrame(data)

print(df)
```

- 'NaN' means Not a Number, indicating a missing value

## Detect Missing Values
- Use **isnull()** to identify missing values.

```python
print(df.isnull())
```

```python
# Count missing values in each column
print(df.isnull().sum())
```

## Remove Missing Values
- Use **dropna()** to remove rows with missing values.

```python
clean_df = df.dropna()
print(clean_df)
```

## Fill Missing Values
- Instead of deleting rows, you can replace missing values.

```python
# Replace missing prices with 0
df['Price'] = df['Price'].fillna(0)

# Replace missing product names
df['Product'] = df['Product'].fillna('Unknown')

print(df)
```

## Find Duplicate Rows
- Use **duplicate()** to identify duplicat records.

```python
print(df.duplicated())
```

- If a row is repeated, it wil return **True**

- Use **drop_duplicates()**

```python
df = df.drop_duplicates()
print(df)
```

- This keeps only unique records

## Check Data Information
- Use **info()** to understand your dataset.

```python
print(df.info())
```

## Generate Summary Statistics
- Use **describe()** to get statistical information.

```python
print(df.describe())
```

# Merge, Join, Concatenate & Pivot Tables with Examples
- In real-world projects, your data is often spread across multiple files or tables.
- Pandas provide powerful functions to combine and summarize data efficiently.

- In this part, you'll learn:
1. Merge DataFrames
2. Join DataFrames
3. Concatenate DataFrames
4. Pivot Tables
5. CrossTab
6. Real-World Use Cases

## Create Sample DataFrames
- We'll use these DataFrames throughout the examples.

```python
import pandas as pd

products = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'Product': ['Laptop', 'Mobile', 'Tablet']
})

sales = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'Sales': [120, 250, 180]
})
```

## Merge DataFrames
- Use "merge()" to combine DataFrames based on a common column.

```python
result = pd.merge(products, sales, on='ProductID')
print(result)
```

## Join DataFrames
- Use "join()" to combine DataFrames using their indexes.

```python
result = products.join(sales['Sales'])
print(result)
```

## Concatenate DataFrames
- Use "concat()" to combine DataFrames vertically.

```python
first = pd.DataFrame({
    'Value': [10, 20]
})
second = pd.DataFrame({
    'Value': [30, 40]
})
result = pd.concat([first, second])
print(result)
```

## Create a Pivot Table
- A Pivot Table summarizes data quickly.

```python
data = pd.DataFrame({
    'Department': ['Sales', 'Sales', 'IT', 'IT'],
    'Salary': [50000, 55000, 70000, 75000]
})
pivot = data.pivot_table(
    values='Salary',
    index='Department',
    aggfunc='mean'
)
print(pivot)
```

## Create a Crosstab
- A crosstab counts the frequency of combinations between columns.

```python
data = pd.DataFrame({
    'Department': ['Sales', 'Sales', 'IT', 'HR'],
    'Shift': ['Morning', 'Evening', 'Morning', 'Evening']
})
print(pd.crosstab(data['Department'], data['Shift']))
```

**Pro Tip**
- Remember this rule:
1. 'merge()' - Combine using common columns
2. 'join()' - Combine using indexes
3. 'concat()' - Stack DataFrame vertically
4. 'pivot_table()' - Summarize data
5. 'crosstab()' - Count combinations

```python
df1 = pd.DataFrame({'A': [1, 2]})
df2 = pd.DataFrame({'B': [3, 4]})

print(pd.concat([df1, df2]).shape)
```

## Working with Dates
- Convert a column to DateTime format using **to_datetime()**.

```python
import pandas as pd

data = {
    'OrderDate': ['2026-08-01', '2026-08-02', '2026-08-03']
}

df = pd.DataFrame(data)

df['OrderDate'] = pd.to_datetime(df['OrderDate'])

df['OrderDate']
```

## Extract Date Components
- Extract useful information from a date.

```python
print(df['OrderDate'].dt.year)
print(df['OrderDate'].dt.month)
print(df['OrderDate'].dt.day)
print(df['OrderDate'].dt.day_of_week)
```

## String Operations
- Use the **.str** accessor to work with text columns.

```python
data = {
    'Product': ['Laptop', 'mobile', 'TABLET']
}

df = pd.DataFrame(data)

print(df['Product'].str.upper())
print(df['Product'].str.lower())
```

## Replace Text
- Replace specific words or characters.

```python
data = {
    'Category': ['Old', 'New', 'Old']
}
df = pd.DataFrame(data)

df['Category'] = df['Category'].str.replace('Old', 'Updated')

print(df)
```

## Apply Custom Functions
- Use **apply()** to perform custom operations.

```python
data = {
    'Price': [100, 250, 400]
}
df = pd.DataFrame(data)

df['Tax'] = df['Price'].apply(lambda price: (price * 0.18)/2)

print(df)
```

## Map Values 
- Replace values using a dictionary

```python
data = {
    'Status': [1, 0, 1, 1, 0]
}
df = pd.DataFrame(data)

status_map = {
    1: 'Active',
    0: 'Inactive'
}

df['Status'] = df['Status'].map(status_map)

print(df)
```

## Replace Values
- Replace values directly.

```python
data = {
    'Priority': ['Low', 'Medium', 'Low', 'High']
}
df = pd.DataFrame(data)
df['Priority'] = df['Priority'].replace('Low', 'Normal')

print(df)
```

## Horizontal Bar Charts  
You can create horizontal bars using 'barh()'.

## What is a Scatter Plot?  
A scatter plot displays individual data points and helps you understand the relationship between two numerical variables.

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
scores = [45, 50, 60, 70, 85]

plt.scatter(hours, scores)

plt.title('Study Hours vs Score')
plt.xlabel('Study Hours')
plt.ylabel('Score')

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# We generate an ARRAY of random numbers
nums = np.random.randint(0, 25, size=100)

# Define bins and labels
bins = [0, 6, 12, 17, 21, 24]
labels = ['Late Night', 'Morning', 'Afternoon', 'Evening', 'Night']

# Categorize the random numbers into your labels
categories = pd.cut(nums, bins=bins, labels=labels, include_lowest=True, right=True)

# Create a DataFrame for easy plotting with Seaborn
df = pd.DataFrame({'Study Hours': nums, 'Time of Day': categories})

# Plot the counts of each category
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Time of Day', order=labels, palette='viridis', hue='Time of Day', legend=False)
plt.title('Distribution of Study Sessions by Time of Day')
plt.ylabel('Number of Sessions')
plt.show()
```

```python
## Scatter plot (Study Hours vs Test Scores)
```

```python
# Generate paired data
study_hours = np.random.randint(0, 24, 100)
# Creating a fake second variable (e.g., test scores) for the y axis
test_scores = study_hours * 2 + np.random.randint(-5, 6, 100)

# Define bins and labels
bins = [0, 6, 12, 17, 21, 24]
labels = ['Late Night', 'Morning', 'Afternoon', 'Evening', 'Night']

# Assign categories based on study hours
time_of_day = pd.cut(study_hours, bins=bins, labels=labels, include_lowest=True)

# Create DataFrame
df = pd.DataFrame({
    'Study Hours': study_hours,
    'Test Score': test_scores,
    'Time of Day': time_of_day
})

# Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Study Hours', y='Test Score', hue='Time of Day', palette='deep')
plt.title('Test Scores vs Study Hours by Time of Day')
plt.show()
```

- Use **pd.cut**: When you want to segment continous data (like hours 0-24) into categorical labels (like 'Morning', 'Evening'), pd.cut is the correct too.
- DataFrames: Passing a Pandas DataFrame to Seaborn (**data=df**) is much cleaner than passing individual lists, especially when using **hue**.

## What are Subplots?  
Subplots allow you to display multiple charts inside a single figure.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 120, 150, 180]
expenses = [60, 70, 80, 90]

plt.subplot(2, 1, 1)
plt.plot(months, sales)
plt.title('Sales')

plt.subplot(2, 1, 2)
plt.plot(months, expenses)
plt.title('Expenses')

plt.tight_layout()
plt.show()
```

subplot(2, 1, 1) means:
- 2 rows
- 1 column
- First chart

## Using plt.subplots()  
A more flexible approach is **plt.subplots()**.

- axes -> Individual chart areas

## Create a DataFrame

```python
# Let's start with a simple dataset
import pandas as pd

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [120, 150, 180, 160, 210],
    'Expenses': [80, 90, 100, 95, 110]
}

df = pd.DataFrame(data)
print(df)
```

## Create a Line Chart
You can use Pandas to plot data with Matplotlib.

```python
import matplotlib.pyplot as plt

df.plot(
    x='Month',
    y='Sales',
    kind='line'
)

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

This is useful for identifying trends. 

## Create a Bar Chart
Compare sales across different months

```python
df.plot(
    x='Month',
    y='Sales',
    kind='bar'
)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

Bar charts make category comparison easier.

## Compare Sales and Expenses
Plot multiple columns together.

```python
df.plot(
    x='Month',
    y=['Sales', 'Expenses'],
    kind='bar'
)
plt.title('Sales vs Expenses')
plt.xlabel('Month')
plt.ylabel('Amount')

plt.show()
```

This allows you to compare two metrics across the same categories.

## Create a Histogram
You can visualize the distribution of a numerical column.

```python
df['Sales'].plot(
    kind='hist',
    bins=5
)
plt.title('Sales Distribution')
plt.xlabel('Sales')

plt.show()
```

Histograms help you understand how numerical values are distributed.

## GroupBy & Visualization
Suppose you have sales data by department.

```python
import pandas as pd

data = {
    'Department': ['Sales', 'Sales', 'IT', 'IT', 'HR', 'HR'],
    'Revenue': [100, 150, 200, 180, 80, 90]
}
df = pd.DataFrame(data)

# Calculate total revenue by department:
summary = df.groupby('Department')['Revenue'].sum().reset_index()

print(summary)
```

```python
# Then visualize it:
summary.plot(kind='bar', x='Department')

plt.title('Revenue by Department')
plt.xlabel('Department')
plt.ylabel('Revenue')

plt.show()
```

- This is a very common Data Analyst workflow:
  Raw Data -> GroupBy -> Summary -> Visualization

## Time-Series Visualization
Dates are extremely important in real-world analytics

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Date': [
        '2026-01-01',
        '2026-02-01',
        '2026-03-01',
        '2026-04-01'
    ],
    'Sales': [100, 230, 160, 190]
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

df.plot(
    x='Date',
    y='Sales',
    kind='line'
)
plt.title('Sales Trend')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.grid(True)

plt.show()
```

This can help identify:
- Growth
- Decline
- Seasonal patterns
- Sudden changes
