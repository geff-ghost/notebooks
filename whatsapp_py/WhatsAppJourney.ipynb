## What is Numpy?  
- Numpy Numerical Python is a powerful Python library used for working with numbers and arrays.

## Why Use Numpy?
- Python lists work well, but Numpy arrays are much faster for mathematical operations
- Numpy is used for:
1. Data Analysis
2. Machine Learning
3. Data Visualization
4. Scientific Computing

```python
# Importing the library
import numpy as np
```

```python
# Creating numpy arrays
numbers = np.array([10, 20, 30, 40])
print(numbers)
```

```python
# Check Array Properties
print(numbers.ndim) # Number of dimentions
print(numbers.size) # Total number of elements
print(numbers.shape) # Structure of the array
```

```python
# Accessing Array Elements
print(numbers[0])
print(numbers[2])
```

```python
# Array Slicing
print(numbers[1:4])
```

```python
# Basic Array Operations
print(numbers + 5)
print(numbers * 2)
```

- Numpy performs operations on every element at once (elementwise operation)

```python
# Print the first and last element
print(numbers[-1])
print(numbers[0])

# Add 100 to every element
print(numbers + 100)
```

## Master 2D Arrays, Reshaping & Mathematical Functions  

### What is a 2D Array?
- A 2D array stores data in rows and columns, just like a table or spreadsheet

```python
import numpy as np

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(matrix)
```

```python
# Check Array Shape
# Use the .shape attribute to find the dimensions of an array

print(matrix.shape) # 2 row and 3 columns
```

```python
# Reshape an Array
# Convert a 1D array into a 2D array using .reshape()
numbers = np.array([1, 2, 3, 4, 5, 6])

new_array = numbers.reshape(2, 3)
print(new_array)
```

- The total number of elements must remain the same

## Mathematical Functions

```python
numbers = np.array([10, 20, 30, 40])

print(np.sum(numbers))
print(np.mean(numbers))
```

## Indexing, Slicing, Boolean Indexing and Filtering
- One of the biggest advantages of NumPy is how quickly you can access and filter data.

### Array Indexing
- Indexing is used to access a specific element in an array.

```python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50, 60])

# Indexing starts from 0
print(numbers[0])
print(numbers[3])
```

### Array Slicing
- Slicing extracts a portion of an array

```python
print(numbers[1:5])
print(numbers[:3])
print(numbers[3:])
print(numbers[-2:])
```

### Indexing in 2D Arrays
- Access elements using row and column indexes.

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix[1, 2])
```

```python
# Access an entire row:
print(matrix[0])

# Access an entire column
print(matrix[:, 1])
```

### Boolean Indexing
- Boolean indexing selects elements that satisfy a condition.

```python
numbers = np.array([12, 25, 8, 40, 15])
print(numbers > 20)
```

### Filtering Data
- Use Boolean indexing to filter values.

```python
print(numbers[numbers > 20])
```

```python
print(numbers[numbers % 2 == 0])
```

### Multiple Conditions
- Filter using more than one condition.

```python
numbers = np.array([12, 25, 8, 40, 15, 30])
result = numbers[(numbers >= 15) & (numbers <= 30)]
print(result)
```

## Sorting, Searching, Stacking & Spliting Arrays  

### Sorting Arrays
- Numpy provides the **sort()** functions to sort array elements in ascending order.

```python
import numpy as np

numbers = np.array([40, 10, 30, 20])

print(np.sort(numbers))
```

### Searching Elements
- Use **where()** to find the index of an element.

```python
numbers = np.array([10, 20, 30, 20, 40])

print(np.where(numbers == 20))
```

This means the value 20 is present at index 1 and 3.

### Stacking Arrays Vertically
- Combine arrays column-wise using **vstack()**.

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

# combines two arrays row-wise
print(np.vstack((array1, array2)))
```

### Stacking Arrays Horizontally
- Combine arrays column-wise using **hstack()**.

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print(np.hstack((array1, array2)))
```

### Splitting Arrays
- Use **split()** to divide an array into equal parts.

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

print(np.split(numbers, 3))
```

### Copy vs View
#### Copy
- Changes to the original array do not affect the copied array.

```python
numbers = np.array([10, 20, 30])

copy_array = numbers.copy()

numbers[0] = 100

print(copy_array)
print(numbers)
```

#### View
- A view shares the same data as the original array.

```python
numbers = np.array([10, 20, 30])

view_array = numbers.view()

numbers[0] = 100
print(view_array)
```

### Find Unique Values
- Use **unique()** to remove duplicate values.

```python
numbers = np.array([10, 20, 20, 30, 30, 40])
print(np.unique(numbers))
```

```python
# Find Index of a specific value
print(np.where(numbers == 20))
```

## Random Numbers, Matrix Operations & Saving Arrays
### Random Number Generation
- NumPy provides the 'random' module to generate random values

```python
import numpy as np
number = np.random.randint(1, 101)
print(number)
```

```python
# Generate Multiple Random Numbers
numbers = np.random.randint(1, 11, size=5)
print(numbers)
```

### Create a Matrix
- A matrix is a two-dimensional NumPy array.

```python
matrix = np.array([
    [1, 2],
    [3, 4]
])
print(matrix)
print('This is a {} matrix'.format(matrix.shape))
```

### Transpose of a Matrix
- Transpose swaps rows and columns.

```python
print(matrix)
print()
print(matrix.T)
```

### Matrix Multiplication
- Use the "@" operator or "np.matmul()".

```python
matrix1 = np.array([
    [1, 2],
    [3, 4]
])
matrix2 = np.array([
    [5, 6],
    [7, 8]
])
print(matrix1 @ matrix2)
```

### Save an Array
- Save a NumPy array to a file using 'save()'.

```python
numbers = np.array([10, 20, 30, 40])
np.save('numbers.npy', numbers)
```

### Load an Array
- Read a saved array using 'load()'

```python
numbers = np.load('numbers.npy')
print(numbers)
```
